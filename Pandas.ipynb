```python
# Create a "Seats" column where every row has a value of 5
carsales["Seats"] = 5
carsales
```

```python
# Create a column called "Engine Size" with random values between 1.3 and 4.5
# Remember: If you're doing it from a Python list, the list has to be the same length
# as the DataFrame
engine_sizes = [1.3, 4.3, 2.3, 3.3, 3.0, 2.3, 1.4, 1.7, 2.5, 3.1]
carsales["Engine Size"] = engine_sizes
carsales
```

```python
# Create a column which represents the price of a car per kilometer
# Then view the DataFrame
carsales["Price per KM"] = carsales["Price"] / carsales["Odometer (KM)"]
carsales
```

```python
# Remove the last column you added using .drop()
carsales = carsales.drop("Price per KM", axis=1)
carsales
```

```python
# Shuffle the DataFrame using sample() with the frac parameter set to 1
# Save the the shuffled DataFrame to a new variable
car_sales_sampled = car_sales_sampled = carsales.sample(frac=1)
# frac 1 -> 100% which mean shuffle all the rows (range from 0 - 1)
car_sales_sampled
```

```python
car_sales_sampled.reset_index(drop=True, inplace=True)
```

```python
# Change the Odometer values from kilometers to miles using a Lambda function
# Then view the DataFrame
carsales["Odometer (KM)"] = carsales["Odometer (KM)"].apply(lambda x: x/1.6)
carsales
```

```python
# Change the title of the Odometer (KM) to represent miles instead of kilometers
carsales = carsales.rename(columns={"Odometer (KM)": "Odometer (Miles)"})
carsales
```
